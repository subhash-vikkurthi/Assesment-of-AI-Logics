### Problem 2 Longest Stable Signal Window

You are monitoring a sensor that records integer signal values at fixed time intervals.
A signal window is considered stable if the difference between the maximum and minimum signal
value inside that window is at most k.
Your task is to find the length of the longest consecutive stable window.
If multiple windows have the same maximum length, return the starting position of the earliest
window.

```python
from collections import deque
```

```python
def solve_str(s):
    data = s.strip().split()
    if not data:
        return "0 0"
    n = int(data[0])
    vals = list(map(int, data[1:]))
    arr = vals[:n]
    k = vals[n] if len(vals) > n else 0

    min_q = deque()
    max_q = deque()
    left = 0
    best_len = 0
    best_start = 0
## check all windows starting from i
    for right in range(len(arr)):
        while max_q and arr[max_q[-1]] <= arr[right]:
            max_q.pop()
        max_q.append(right)

        while min_q and arr[min_q[-1]] >= arr[right]:
            min_q.pop()
        min_q.append(right)

        while arr[max_q[0]] - arr[min_q[0]] > k:
            if max_q[0] == left:
                max_q.popleft()
            if min_q[0] == left:
                min_q.popleft()
            left += 1

        cur = right - left + 1
        if cur > best_len:
            best_len = cur
            best_start = left

    return f"{best_len} {best_start + 1}"
```

### Problem 3 Longest Stable Signal Window

You are given two non-empty linked lists representing two non-negative integers. The digits are
stored in reverse order, and each node contains a single digit.
Add the two numbers and return the sum as a linked list in the same reverse-order format.
You may assume the two numbers do not contain leading zeros, except the number 0 itself.

```python
def solve(test_case):
    ### Remove Spaces
    data = test_case.strip().split()
    n = int(data[0])
    l1 = list(map(int, data[1:1+n]))
    m = int(data[1+n])
    l2 = list(map(int, data[2+n:2+n+m]))

    carry=0
    i=j=0
    ### TO Store 
    ans=[]
    while i<len(l1) or j<len(l2) or carry:
        a = l1[i] if i<len(l1) else 0
        b = l2[j] if j<len(l2) else 0
        s=a+b+carry
        ans.append(s%10)
        carry=s//10
        if i<len(l1): i+=1
        if j<len(l2): j+=1

    return ans
```
